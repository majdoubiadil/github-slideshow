```python
import pandas as pd
import numpy as np
df = pd.read_excel("wpt_rectenna_dataset.xlsx")
df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.hist(df["Frequency (GHz)"], bins=20)

plt.title("Distribution de la fréquence")
plt.xlabel("Fréquence (GHz)")
plt.ylabel("Nombre d'échantillons")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(df["RF-to-DC Efficiency (%)"], bins=20)

plt.title("Distribution de l'efficacité RF-to-DC")
plt.xlabel("Efficacité (%)")
plt.ylabel("Nombre d'échantillons")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.boxplot(df["RF-to-DC Efficiency (%)"])

plt.title("Boxplot de l'efficacité RF-to-DC")

plt.ylabel("Efficacité (%)")

plt.show()
```

```python
import seaborn as sns
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")
plt.title("Matrice de corrélation")
plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.scatter(df["Input Power (dBm)"],
            df["RF-to-DC Efficiency (%)"])

plt.xlabel("Puissance d'entrée (dBm)")
plt.ylabel("Efficacité RF-to-DC (%)")

plt.title("Puissance d'entrée vs efficacité")

plt.grid(True)

plt.show()
```
